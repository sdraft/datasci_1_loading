Issuers required to submit a Progress Report form include issuers that have previously submitted an Implementation Plan form, have continued to oeprate on an FFE, and have met the enrollemnt requirement of greater than 500 enrollees as of July 1 of 2025, when applicable. Enrollment is evaluated every two years beginning on the third year of progress reporting.

**Number of Progress Report Submissions Required** - indicator of the number of Progress Report submissions required. Issuers that are tracking multiple Quality Improvement Strategies will be required to submit a Progress Report for each QIS the issuer is tracking.

###Features

**:id** -	unique system-generated identifier for the record

**:version** - system-generated version identifier for the record

**:created_at** -	date and time the record was created

**:updated_at** -	date and time the record was last updated

**year** -	year associated with the health measure

**location** -	geographic location represented, such as a state or the United States

**short_text_indicator** -	short name of the health condition or indicator

**long_text_indicator** -	full description of the health measure being reported

**stratification1** -	primary demographic
stratification category, such as Sex

**stratification_group2** -	specific group within the secondary stratification, such as 65-74

**stratification_group2_label** -	display label corresponding to the secondary stratification group

**data_type** -	type of measurement reported, such as Total Cases or Crude Rate per 100

**data_type_label** -	display-friendly label for the type of measurement

**data_value** -numeric value associated with the health measure

**fips** -	Federal Information Processing Standards code identifying the geographic area

**geolocation** -	geographic coordinates associated with the location

**:@computed_region_hjsp_umg2** -	system-generated geographic region identifier used by the data platform

**:@computed_region_skr5_azej** -	additional system-generated geographic region identifier

**_95ci** - reported 95% confidence interval for the estimate, when available

**lower_confidence_limit** - lower bound of the confidence interval

**upper_confidence_limit** - upper bound of the confidence interval
